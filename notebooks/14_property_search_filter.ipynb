```python
# 🔎 Property Filtering Engine

The following function allows multiple filters to be applied simultaneously.

Available filters:

- Locality
- Minimum price
- Maximum price
- Minimum area
- Maximum area
- Bedrooms
- Bathrooms
- Property type
- Furnishing
- Minimum price/sq.ft.
- Maximum price/sq.ft.
- Deal status
- Anomaly status
```

```python
def search_properties(
    data,
    locality=None,
    min_price=None,
    max_price=None,
    min_area=None,
    max_area=None,
    bedrooms=None,
    bathrooms=None,
    property_type=None,
    furnished=None,
    min_ppsf=None,
    max_ppsf=None,
    deal_status=None,
    anomaly_status=None
):

    result = data.copy()

    # Locality
    if locality is not None:

        if locality != "All":

            result = result[
                result["locality"].astype(str).str.lower()
                == str(locality).lower()
            ]

    # Price
    if min_price is not None:

        result = result[
            result["price"] >= min_price
        ]

    if max_price is not None:

        result = result[
            result["price"] <= max_price
        ]

    # Area
    if min_area is not None:

        result = result[
            result["area"] >= min_area
        ]

    if max_area is not None:

        result = result[
            result["area"] <= max_area
        ]

    # Bedrooms
    if bedrooms is not None:

        result = result[
            result["bedroom_num"] == bedrooms
        ]

    # Bathrooms
    if bathrooms is not None:

        result = result[
            result["bathroom_num"] == bathrooms
        ]

    # Property type
    if property_type is not None:

        if property_type != "All":

            result = result[
                result["property_type"].astype(str).str.lower()
                == str(property_type).lower()
            ]

    # Furnishing
    if furnished is not None:

        if furnished != "All":

            result = result[
                result["furnished"].astype(str).str.lower()
                == str(furnished).lower()
            ]

    # Price per sq.ft.
    if min_ppsf is not None:

        result = result[
            result["price_per_sqft"] >= min_ppsf
        ]

    if max_ppsf is not None:

        result = result[
            result["price_per_sqft"] <= max_ppsf
        ]

    # Deal status
    if deal_status is not None:

        if "deal_status" in result.columns:

            result = result[
                result["deal_status"].astype(str).str.lower()
                == str(deal_status).lower()
            ]

    # Anomaly status
    if anomaly_status is not None:

        if anomaly_status == "Normal":

            result = result[
                result["is_anomaly"] == False
            ]

        elif anomaly_status == "Anomaly":

            result = result[
                result["is_anomaly"] == True
            ]

    result = result.reset_index(
        drop=True
    )

    return result


print("✅ Property search function created successfully!")
```

```python
## 💰 Deal Status Classification

We classify properties based on the difference between actual listed/observed price and predicted market price.

### Categories:

- Attractive Deal → 15% or more below predicted value
- Reasonable Deal → 5% to 15% below predicted value
- Close to Market Value → within approximately ±5%
- Slightly Expensive → 5% to 15% above predicted value
- Expensive Deal → more than 15% above predicted value
```

```python
def calculate_deal_status(row):

    if "price_difference_percent" not in row.index:

        return "Not Available"

    difference = row["price_difference_percent"]

    if pd.isna(difference):

        return "Not Available"

    if difference <= -15:

        return "Attractive Deal"

    elif difference <= -5:

        return "Reasonable Deal"

    elif difference < 5:

        return "Close to Market Value"

    elif difference < 15:

        return "Slightly Expensive"

    else:

        return "Expensive Deal"


search_df["deal_status"] = search_df.apply(
    calculate_deal_status,
    axis=1
)

print("Deal status classification completed.")

display(
    search_df[
        [
            "price",
            "predicted_price",
            "price_difference_percent",
            "deal_status"
        ]
    ].head(10)
)
```

```python
# ⭐ Property Ranking

After filtering properties, we can rank them based on:

1. Lower price per sq.ft.
2. Lower price compared with predicted market value
3. Higher area
4. Lower anomaly risk
```

```python
ranking_df = search_df.copy()

ranking_df["value_score"] = 0.0

# Lower price per sq.ft. gets better score
ppsf_rank = ranking_df["price_per_sqft"].rank(
    ascending=True,
    pct=True
)

# Lower price difference gets better score
if "price_difference_percent" in ranking_df.columns:

    difference_rank = ranking_df[
        "price_difference_percent"
    ].rank(
        ascending=True,
        pct=True
    )

else:

    difference_rank = 0.5

# Higher area gets better score
area_rank = ranking_df["area"].rank(
    ascending=False,
    pct=True
)

ranking_df["value_score"] = (
    (1 - ppsf_rank) * 0.40
    +
    (1 - difference_rank) * 0.40
    +
    area_rank * 0.20
)

ranking_df = ranking_df.sort_values(
    "value_score",
    ascending=False
).reset_index(
    drop=True
)

print("Property ranking completed.")

display(
    ranking_df[
        [
            "locality",
            "price",
            "area",
            "price_per_sqft",
            "value_score"
        ]
    ].head(20)
)
ranking_df = search_df.copy()

ranking_df["value_score"] = 0.0

# Lower price per sq.ft. gets better score
ppsf_rank = ranking_df["price_per_sqft"].rank(
    ascending=True,
    pct=True
)

# Lower price difference gets better score
if "price_difference_percent" in ranking_df.columns:

    difference_rank = ranking_df[
        "price_difference_percent"
    ].rank(
        ascending=True,
        pct=True
    )

else:

    difference_rank = 0.5

# Higher area gets better score
area_rank = ranking_df["area"].rank(
    ascending=False,
    pct=True
)

ranking_df["value_score"] = (
    (1 - ppsf_rank) * 0.40
    +
    (1 - difference_rank) * 0.40
    +
    area_rank * 0.20
)

ranking_df = ranking_df.sort_values(
    "value_score",
    ascending=False
).reset_index(
    drop=True
)

print("Property ranking completed.")

display(
    ranking_df[
        [
            "locality",
            "price",
            "area",
            "price_per_sqft",
            "value_score"
        ]
    ].head(20)
)
```

```python
def search_and_rank_properties(
    data,
    locality=None,
    min_price=None,
    max_price=None,
    min_area=None,
    max_area=None,
    bedrooms=None,
    bathrooms=None,
    property_type=None,
    furnished=None,
    min_ppsf=None,
    max_ppsf=None,
    deal_status=None,
    anomaly_status=None,
    top_n=20
):

    results = search_properties(
        data=data,
        locality=locality,
        min_price=min_price,
        max_price=max_price,
        min_area=min_area,
        max_area=max_area,
        bedrooms=bedrooms,
        bathrooms=bathrooms,
        property_type=property_type,
        furnished=furnished,
        min_ppsf=min_ppsf,
        max_ppsf=max_ppsf,
        deal_status=deal_status,
        anomaly_status=anomaly_status
    )

    if len(results) == 0:

        return results

    results = results.copy()

    # Value score
    ppsf_rank = results[
        "price_per_sqft"
    ].rank(
        ascending=True,
        pct=True
    )

    if "price_difference_percent" in results.columns:

        difference_rank = results[
            "price_difference_percent"
        ].rank(
            ascending=True,
            pct=True
        )

    else:

        difference_rank = pd.Series(
            0.5,
            index=results.index
        )

    area_rank = results[
        "area"
    ].rank(
        ascending=False,
        pct=True
    )

    results["value_score"] = (
        (1 - ppsf_rank) * 0.40
        +
        (1 - difference_rank) * 0.40
        +
        area_rank * 0.20
    )

    results = results.sort_values(
        "value_score",
        ascending=False
    )

    results = results.head(
        top_n
    ).reset_index(
        drop=True
    )

    return results


print("✅ Search + ranking function created!")
```
