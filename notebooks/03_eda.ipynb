```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("../data/processed/mumbai_house_price_cleaned.csv")

print("Cleaned dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[-1, 5, 10, 20, 30, np.inf],
    labels=[
        "0-5 years",
        "6-10 years",
        "11-20 years",
        "21-30 years",
        "30+ years"
    ]
)

df["age_group"].value_counts().sort_index()
```

```python
age_group_price = (
    df.groupby(
        "age_group",
        observed=True
    )["price"]
    .agg(["count", "mean", "median"])
)

age_group_price
```

```python
numerical_df = df.select_dtypes(
    include=np.number
)

correlation_matrix = numerical_df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix of Numerical Features")

plt.show()
```

```python
price_correlation = (
    correlation_matrix["price"]
    .sort_values(ascending=False)
)

price_correlation
```

```python
print("Correlation with Price")
print("=" * 40)

for feature, correlation in price_correlation.items():
    print(f"{feature:20} : {correlation:.4f}")
```

```python
selected_features = [
    "price",
    "area",
    "bedroom_num",
    "bathroom_num",
    "age",
    "total_floors",
    "price_per_sqft"
]

sns.pairplot(
    df[selected_features].sample(
        min(3000, len(df)),
        random_state=42
    )
)

plt.show()
```
