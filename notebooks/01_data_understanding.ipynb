```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
df = pd.read_csv("../data/raw/mumbai-house-price-data-cleaned.csv")

print("Dataset loaded successfully!")
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": missing_percentage
})

missing_table.sort_values(
    by="Percentage",
    ascending=False
)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
numerical_columns = df.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns

print("Numerical columns:")
print(numerical_columns.tolist())
```

```python
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

print("Categorical columns:")
print(categorical_columns.tolist())
```

```python
print("Target variable: price")

print("\nPrice statistics:")
print(df["price"].describe())
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["price"],
    bins=50,
    kde=True
)

plt.title("Distribution of House Prices")
plt.xlabel("Price")
plt.ylabel("Number of Properties")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["area"],
    bins=50,
    kde=True
)

plt.title("Distribution of Property Area")
plt.xlabel("Area")
plt.ylabel("Number of Properties")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="property_type"
)

plt.title("Property Type Distribution")
plt.xlabel("Property Type")
plt.ylabel("Number of Properties")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="furnished"
)

plt.title("Furnishing Status Distribution")
plt.xlabel("Furnished Status")
plt.ylabel("Number of Properties")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="bedroom_num"
)

plt.title("Number of Bedrooms Distribution")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Number of Properties")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="bathroom_num"
)

plt.title("Number of Bathrooms Distribution")
plt.xlabel("Number of Bathrooms")
plt.ylabel("Number of Properties")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="area",
    y="price",
    alpha=0.5
)

plt.title("House Price vs Property Area")
plt.xlabel("Area")
plt.ylabel("Price")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="bedroom_num",
    y="price"
)

plt.title("House Price by Number of Bedrooms")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Price")

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="property_type",
    y="price"
)

plt.title("House Price by Property Type")
plt.xlabel("Property Type")
plt.ylabel("Price")

plt.xticks(rotation=45)

plt.show()
```

```python
top_localities = df["locality"].value_counts().head(15)

print("Top 15 localities by number of properties:")
print(top_localities)
```

```python
plt.figure(figsize=(12, 7))

top_localities.sort_values().plot(
    kind="barh"
)

plt.title("Top 15 Localities by Number of Properties")
plt.xlabel("Number of Properties")
plt.ylabel("Locality")

plt.show()
```

```python
avg_price_property_type = (
    df.groupby("property_type")["price"]
    .mean()
    .sort_values(ascending=False)
)

print(avg_price_property_type)
```

```python
avg_price_furnished = (
    df.groupby("furnished")["price"]
    .mean()
    .sort_values(ascending=False)
)

print(avg_price_furnished)
```

```python
avg_price_bedrooms = (
    df.groupby("bedroom_num")["price"]
    .mean()
    .sort_index()
)

print(avg_price_bedrooms)
```

```python
correlation_matrix = df.select_dtypes(
    include=np.number
).corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
price_correlation = (
    correlation_matrix["price"]
    .sort_values(ascending=False)
)

print("Correlation of numerical features with price:")
print(price_correlation)
```
